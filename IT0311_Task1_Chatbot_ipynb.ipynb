```python
# ==========================================================
# Import Required Libraries
# ==========================================================

import gradio as gr
import random
from datetime import datetime
```

```python
# ==========================================================
# Define Chatbot Intents and Responses
# ==========================================================

# Greeting Intent
greetings = [
    "hello",
    "hi",
    "hey",
    "good morning",
    "good afternoon",
    "good evening"
]

greeting_responses = [
    "Hello! Welcome to the IT0311 AI Chatbot.",
    "Hi there! How can I assist you today?",
    "Greetings! Feel free to ask me anything about AI or IoT."
]

# Goodbye Intent
goodbyes = [
    "bye",
    "goodbye",
    "see you",
    "farewell"
]

goodbye_responses = [
    "Goodbye! Have a wonderful day.",
    "See you next time!",
    "Take care and keep learning!"
]

# Help Intent
help_keywords = [
    "help",
    "support",
    "assist"
]

# Feedback Intent
feedback_keywords = [
    "feedback",
    "comment",
    "suggestion"
]

# Thank You Intent
thanks_keywords = [
    "thanks",
    "thank you",
    "appreciate"
]
```

```python
# ==========================================================
# Chatbot Function
# ==========================================================

def chatbot(message, history):

    # Convert user input to lowercase
    message = message.lower()

    # -----------------------------
    # Greeting Intent
    # -----------------------------
    if any(word in message for word in greetings):
        return random.choice(greeting_responses)

    # -----------------------------
    # Goodbye Intent
    # -----------------------------
    elif any(word in message for word in goodbyes):
        return random.choice(goodbye_responses)

    # -----------------------------
    # Help Intent
    # -----------------------------
    elif any(word in message for word in help_keywords):
        return (
            "I can help you with:\n\n"
            "• Greetings\n"
            "• Questions about AI\n"
            "• Questions about IoT\n"
            "• Current time\n"
            "• Weather information\n"
            "• IT0311 course information\n"
            "• University contact details\n"
            "• Feedback\n"
            "• Saying goodbye"
        )

    # -----------------------------
    # Feedback Intent
    # -----------------------------
    elif any(word in message for word in feedback_keywords):
        return (
            "Thank you for your feedback! "
            "It helps improve this chatbot."
        )

    # -----------------------------
    # Thank You Intent
    # -----------------------------
    elif any(word in message for word in thanks_keywords):
        return "You're welcome! Happy to help."

    # -----------------------------
    # AI Questions
    # -----------------------------
    elif "what is ai" in message:

        return (
            "Artificial Intelligence (AI) is a branch of computer science "
            "that enables computers to perform tasks that normally require "
            "human intelligence, such as learning, reasoning, and decision-making."
        )

    # -----------------------------
    # IoT Questions
    # -----------------------------
    elif "what is iot" in message:

        return (
            "The Internet of Things (IoT) is a network of smart devices "
            "that communicate and exchange data over the Internet."
        )

    # -----------------------------
    # Current Time
    # -----------------------------
    elif "time" in message:

        current_time = datetime.now().strftime("%I:%M %p")

        return f"The current time is {current_time}."

    # -----------------------------
    # Weather
    # -----------------------------
    elif "weather" in message:

        return (
            "I cannot access live weather information. "
            "However, today's weather can be checked using Google Weather "
            "or your preferred weather application."
        )

    # -----------------------------
    # Course Information
    # -----------------------------
    elif "it0311" in message or "course" in message:

        return (
            "IT0311 is the Introduction to Artificial Intelligence and IoT course. "
            "Students learn AI concepts, machine learning, chatbots, "
            "sentiment analysis, image classification, and IoT fundamentals."
        )

    # -----------------------------
    # Lecturer
    # -----------------------------
    elif "lecturer" in message:

        return (
            "Please refer to your course outline or the university portal "
            "for your lecturer's information."
        )

    # -----------------------------
    # University
    # -----------------------------
    elif "ibs" in message or "university" in message:

        return (
            "IBSUniversity offers programs in Business, Accounting, "
            "Information Technology, and other professional disciplines."
        )

    # -----------------------------
    # Contact Details
    # -----------------------------
    elif "contact" in message:

        return (
            "Please visit the IBSUniversity website or contact the Student Services Office "
            "for official contact information."
        )

    # -----------------------------
    # Chatbot Name
    # -----------------------------
    elif "your name" in message:

        return (
            "I am the IT0311 AI Chatbot developed for Assignment 1."
        )

    # -----------------------------
    # Default Response
    # -----------------------------
    else:

        return (
            "Sorry, I don't understand your question.\n\n"
            "Try asking about:\n"
            "• AI\n"
            "• IoT\n"
            "• Weather\n"
            "• Time\n"
            "• IT0311\n"
            "• University\n"
            "• Help"
        )
```

```python
print(chatbot("Hello", []))
print(chatbot("help", []))
print(chatbot("thank you", []))
print(chatbot("what is ai", []))
print(chatbot("bye", []))
```

```python
# ==========================================================
# Create Chat Interface
# ==========================================================

demo = gr.ChatInterface(
    fn=chatbot,
    title="🤖 IT0311 AI Chatbot",
    description="""
Welcome to the IT0311 Artificial Intelligence and IoT Chatbot.

You can ask me about:
• Artificial Intelligence
• Internet of Things
• Current Time
• Weather
• IT0311 Course
• University Information
• Feedback
"""
)

```

```python
# ==========================================================
# Launch Chatbot
# ==========================================================

demo.launch()
```
